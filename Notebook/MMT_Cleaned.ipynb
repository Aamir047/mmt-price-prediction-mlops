```python
df["Package Type"].value_counts().tail(20)

#Counts how many times each value appears in the Package Type column.
#Show me the 20 least frequent Package Type values so I can inspect the suspicious ones.
```

```python
#This code checks every value in the Package Type column and keeps only the five valid categories: 
# Budget, Standard, Deluxe, Premium, and Luxury. The ~ means "not", so it identifies values that 
# are not in that list, and df.loc[...] selects those rows. 
# Finally, = np.nan replaces those invalid URL-like values with missing values (NaN) 
# instead of deleting the entire rows, allowing us to handle them later during preprocessing.


df.loc[
    ~df["Package Type"].isin(
        ["Budget", "Standard", "Deluxe", "Premium", "Luxury"]
    ),
    "Package Type"
] = np.nan
```

```python
#counts how many times each value appears in the Package Type column.
#By default, value_counts() doesn't show missing (NaN) values. 
# Setting dropna=False tells pandas to include them in the count, so we can see how many rows have missing values in the Package Type column after our cleaning step.

df["Package Type"].value_counts(dropna=False)
```

```python
df["Package Type"] = df["Package Type"].fillna(
    df["Package Type"].mode()[0]
)
```

```python
df["Package Type"].isnull().sum()
```

```python
#Now we will handle the outliers in price per person column.
#As we checked earlier it was skewed by 2.57 

df["Per Person Price"].skew()
```

```python
Q1 = df["Per Person Price"].quantile(0.25)
Q3 = df["Per Person Price"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Lower:", lower)
print("Upper:", upper)
```

```python
#This is legit packages with price per person greater than upper limit.
#Hence useful

df[df["Per Person Price"] > upper].shape[0]
```
